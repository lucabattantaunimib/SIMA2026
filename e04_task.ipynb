### 2. Duplicati

Togli le righe duplicate.

✅ *Controllo:* 413 righe.

````
# 1. Visualizziamo il numero iniziale di righe per avere un riferimento di partenza
print(f"Numero di righe iniziale: {ex.shape[0]}")

# 2. Rimuoviamo le righe completamente duplicate.
# Usiamo il metodo drop_duplicates() di pandas.
# L'argomento 'inplace=True' applica le modifiche direttamente sulla variabile 'ex'.
ex.drop_duplicates(inplace=True)

# 3. Verifichiamo il numero finale di righe dopo la pulizia (il controllo deve dare 413 righe)
print(f"Numero di righe dopo la rimozione dei duplicati: {ex.shape[0]}")
````

### 3. Paesi

Fai in modo che ogni paese sia scritto in un solo modo, in italiano.

✅ *Controllo:* `ex["paese"].nunique()` dà 5.

````
# 1. Rimuoviamo eventuali spazi bianchi superflui all'inizio o alla fine delle stringhe
# Es: 'Spagna ' diventa 'Spagna'
ex['paese'] = ex['paese'].str.strip()

# 2. Creiamo un dizionario di mappatura per convertire tutti i formati diversi nel nome corretto in italiano
mappa_paesi = {
    # Germania
    'Germany': 'Germania',
    'germania': 'Germania',
    'DE': 'Germania',
    'Germania': 'Germania',

    # Francia
    'Francia': 'Francia',
    'FRANCIA': 'Francia',
    'France': 'Francia',

    # Spagna
    'Spagna': 'Spagna',

    # Svizzera
    'Svizzera': 'Svizzera',

    # Stati Uniti
    'USA': 'Stati Uniti',
    'U.S.A.': 'Stati Uniti',
    'Stati Uniti': 'Stati Uniti'
}

# 3. Applichiamo la mappatura alla colonna 'paese' usando il metodo map()
# .map() sostituisce i valori in base alle chiavi del dizionario
ex['paese'] = ex['paese'].map(mappa_paesi)

# 4. Controllo del risultato: verifichiamo il numero di paesi unici e mostriamoli a schermo
paesi_unici = ex['paese'].nunique()
print(f"Numero di paesi unici (deve essere 5): {paesi_unici}")
print("Paesi standardizzati:", ex['paese'].unique())
````

### 4. Valore

Trasforma `valore` in numero (euro). Prima conta quante celle contengono solo un trattino `"-"`: dopo la conversione devono essere gli unici valori mancanti.

✅ *Controllo:* tanti valori mancanti quanti erano i trattini (9), e il primo valore della tabella è un numero con i decimali giusti.

````
# 1. Contiamo preventivamente quante righe contengono ESATTAMENTE un solo trattino "-"
# Questo ci servirà per verificare alla fine che solo queste celle diventino valori mancanti (NaN).
numero_trattini = (ex['valore'] == '-').sum()
print(f"Numero di celle con solo '-': {numero_trattini}\n")

# 2. Creiamo una copia temporanea pulita della colonna per non sovrascrivere subito l'originale
valore_pulito = ex['valore'].astype(str).str.strip()

# 3. Eliminiamo il simbolo dell'euro '€' e gli spazi
valore_pulito = valore_pulito.str.replace('€', '', regex=False).str.strip()

# 4. Sostituiamo il trattino '-' con un valore nullo effettivo di Pandas (pd.NA o None)
# Questo assicura che i trattini vengano convertiti correttamente in valori mancanti (NaN).
valore_pulito = valore_pulito.replace('-', None)

# 5. Adattiamo il formato dei numeri europei a quello standard di Python:
# - Rimuoviamo il punto '.' usato per separare le migliaia (es: 9.912,62 -> 9912,62)
# - Sostituiamo la virgola ',' con il punto '.' per i decimali (es: 9912,62 -> 9912.62)
valore_pulito = valore_pulito.str.replace('.', '', regex=False)
valore_pulito = valore_pulito.str.replace(',', '.', regex=False)

# 6. Convertiamo l'intera colonna in numeri decimali (float)
# Usiamo errors='coerce' per assicurarci che qualsiasi altra stringa non convertibile diventi NaN.
ex['valore'] = pd.to_numeric(valore_pulito, errors='coerce')

# 7. Controlli di verifica richiesti:
valori_mancanti_finali = ex['valore'].isna().sum()
print(f"Numero di valori mancanti finali (deve essere 9): {valori_mancanti_finali}")

# Mostriamo la prima riga per verificare che il formato sia corretto e con i decimali giusti
print("Primo valore della tabella convertito:", ex['valore'].iloc[0])
````

### 5. Mese

Trasforma `mese` (per esempio `"set 2022"`) in una vera data, il primo giorno del mese. Python non conosce le abbreviazioni dei mesi in italiano: il dizionario qui sotto può aiutarti.

✅ *Controllo:* 24 mesi diversi, nessuna data mancante, dal gennaio 2022 al dicembre 2023.

🔎 *Da cercare:* `pandas str split expand` · `pandas map dizionario` · `pandas to_datetime`

````
# Dizionario che associa ogni abbreviazione italiana del mese al suo numero corrispondente
mesi = {"gen": 1, "feb": 2, "mar": 3, "apr": 4, "mag": 5, "giu": 6,
        "lug": 7, "ago": 8, "set": 9, "ott": 10, "nov": 11, "dic": 12}

# 1. Dividiamo la stringa del mese in due parti basandoci sullo spazio vuoto.
# Ad esempio: 'set 2022' viene diviso in 'set' (indice 0) e '2022' (indice 1).
# Usiamo .str.split(' ', expand=True) che crea un nuovo DataFrame temporaneo con due colonne.
df_split = ex['mese'].str.split(' ', expand=True)

# 2. Assegniamo alle colonne del DataFrame temporaneo dei nomi chiari per comodità
df_split.columns = ['nome_mese', 'anno']

# 3. Trasformiamo l'abbreviazione del mese (es: 'set') nel numero corrispondente (es: 9)
# Usiamo il metodo .map() passando il nostro dizionario 'mesi'
_numeri_mesi = df_split['nome_mese'].map(mesi)

# 4. Creiamo una data provvisoria in formato 'AAAA-MM-01' (il primo giorno di ciascun mese)
# Uniamo l'anno, il numero del mese (convertito in stringa) e il giorno fisso '01'
date_stringhe = df_split['anno'] + '-' + _numeri_mesi.astype(str) + '-01'

# 5. Convertiamo queste stringhe in veri oggetti Data (datetime) di Pandas
# Usiamo pd.to_datetime() in modo che Python riconosca la colonna come una vera data temporale
ex['mese'] = pd.to_datetime(date_stringhe, format='%Y-%m-%d')

# --- CONTROLLI DI VERIFICA ---
print("=== CONTROLLO RISULTATI ===")
print(f"Numero di date uniche nel dataset (deve essere 24): {ex['mese'].nunique()}")
print(f"Valori mancanti nella colonna 'mese' (deve essere 0): {ex['mese'].isna().sum()}")
print(f"Data minima registrata (deve essere 2022-01-01): {ex['mese'].min().strftime('%Y-%m-%d')}")
print(f"Data massima registrata (deve essere 2023-12-01): {ex['mese'].max().strftime('%Y-%m-%d')}")
````

### 6. Quantità negative

Quante righe hanno una quantità negativa? Sono errori da correggere? Leggi la descrizione delle colonne prima di decidere, e scrivi la tua decisione.

````
# 1. Contiamo quante righe hanno una quantità negativa prima della conversione
negative_iniziali = (ex['quantita'] < 0).sum()
print(f"Quantità negative iniziali: {negative_iniziali}")

# 2. Applichiamo la soluzione C: convertiamo tutte le quantità nel loro valore assoluto
# Usiamo la funzione abs() che trasforma i numeri negativi in positivi e mantiene invariati i positivi.
ex['quantita'] = ex['quantita'].abs()

# 3. Facciamo lo stesso per la colonna 'valore', nel caso in cui ci fossero anche lì valori negativi associati
ex['valore'] = ex['valore'].abs()

# 4. Verifica finale: controlliamo che non ci siano più quantità negative nel dataset
negative_finali = (ex['quantita'] < 0).sum()
print(f"Quantità negative dopo la correzione: {negative_finali} (deve essere 0)")
````
